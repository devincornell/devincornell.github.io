```python
from typing import Annotated
from dataclasses import dataclass
from pydantic import BaseModel, Field
import json

@dataclass(slots=True, frozen=True)
class MyDataclass:
    name: Annotated[str, Field(description="The title of the data class", default='<NO_TITLE>')]
    bio: Annotated[str, Field(description="The description of the data class")]

class MyPydanticType(BaseModel):
    uuid: Annotated[str, Field(description="Unique identifier of the instance.", title="Unique Identifier", gt=-1, default=0)]
    mdc: MyDataclass
print(json.dumps(MyPydanticType.model_json_schema(), indent=2))
```

```python
import json
MyPydanticType.model_validate({'identifier': 1, 'mdc': {'description': 'fuck'}})
```

```python
MyPydanticType(mdc = MyDataclass(title='dammit', description='fuck'))
```

```python
@dataclasses.dataclass(slots=True, frozen=True)
class MyDataClass:
    title: Annotated[str, pydantic.Field(description="The title of the data class")]
    description: Annotated[str, pydantic.Field(description="The description of the data class")]
```

```python
@dataclasses.dataclass(slots=True, frozen=True)
class MyTinyDataClass:
    tiny_field: typing.Annotated[int, pydantic.Field(gt=0, description="A tiny integer field")]

@dataclasses.dataclass(slots=True, frozen=True)
class MyDataClass:
    mdc_field1: int
    mdc_field2: str
    tdc: MyTinyDataClass

class MyPydantic(pydantic.BaseModel):
    pt_field1: typing.Annotated[int, pydantic.Field(description="An integer field", title="PT Field 1")]
    pt_field2: str
    mdc: MyDataClass
print(json.dumps(MyPydantic.model_json_schema(), indent=2))
```

```python
mpd = MyPydantic(field1=1, field2="example", mdc=MyDataClass(field1=1, field2="example", tdc=MyTinyDataClass(tiny_field=1)))
dump = mpd.model_dump()

MyPydantic.model_validate(dump)
```

```python
class Me1:
    name: str
    age: int
    def __init__(self, name: str, age: int):
        self.name = name
        self.age = age

class Me2(pydantic.BaseModel):
    name: str
    age: int

class Me3(pydantic.BaseModel):
    name: str
    age: typing.Annotated[int, pydantic.Field(gt=0, lt=150)]

@dataclasses.dataclass
class Me4:
    name: str
    age: int
```

```python
SIZE = int(1e7)
names_and_ages = [(f'name_{i}', (i % 140) + 1) for i in range(1,SIZE)]
names_and_ages[:5]
```

```python
%timeit [Me1(name=name, age=age) for name, age in names_and_ages]
%timeit [Me2(name=name, age=age) for name, age in names_and_ages]
%timeit [Me3(name=name, age=age) for name, age in names_and_ages]
```
